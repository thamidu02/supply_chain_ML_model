```python
# covareiate shift audit
import numpy as np
from scipy.stats import ks_2samp

# Helper function to compute Population Stability Index (PSI)
def calculate_psi(train_series, test_series, bins=10):
    quantiles = np.linspace(0, 1, bins + 1)
    bin_edges = np.percentile(train_series.dropna(), quantiles * 100)
    bin_edges = np.unique(bin_edges)
    if len(bin_edges) < 2:
        return 0.0
    bin_edges[0] = -np.inf
    bin_edges[-1] = np.inf
    
    train_counts, _ = np.histogram(train_series.dropna(), bins=bin_edges)
    test_counts, _ = np.histogram(test_series.dropna(), bins=bin_edges)
    
    train_pct = np.clip(train_counts / len(train_series.dropna()), 1e-4, 1.0)
    test_pct = np.clip(test_counts / len(test_series.dropna()), 1e-4, 1.0)
    
    psi_val = np.sum((test_pct - train_pct) * np.log(test_pct / train_pct))
    return round(psi_val, 4)


audit_features = [
    'Days for shipment (scheduled)',
    'Sales per customer',
    'Product Price',
    'Benefit per order',
    'Order Item Discount Rate'
]

shift_results = []
for col in audit_features:
    stat, p_val = ks_2samp(train_df[col], test_df[col])
    psi = calculate_psi(train_df[col], test_df[col])
    status = "Stable (PSI < 0.10)" if psi < 0.10 else "Moderate Drift" if psi < 0.25 else "Severe Shift"
    shift_results.append({
        "Feature": col,
        "KS_Statistic": round(stat, 4),
        "p_value": round(p_val, 4),
        "PSI": psi,
        "Drift_Status": status
    })

shift_report_df = pd.DataFrame(shift_results)
print(shift_report_df.to_string(index=False))
```

### Decision Log Entry: Covariate Shift Verification
* **Evidence**: Two-sample Kolmogorov-Smirnov (KS) tests and Population Stability Index (PSI) were calculated across all core numerical features.
* **Finding**: All features yielded $\text{PSI} \le 0.0004$ (drastically below the $0.10$ stability threshold) and KS p-values $> 0.20$.
* **Conclusion**: Proves that the grouped split by `Order Id` preserved identical underlying feature distributions without inducing covariate drift.


```python
from sklearn.ensemble import IsolationForest

num_features = [
    'Days for shipment (scheduled)',
    'Sales per customer',
    'Product Price',
    'Benefit per order',
    'Order Item Discount Rate',
    'Order Item Quantity',
    'Order Item Profit Ratio',
    'Order Item Total'
]

iso_forest=IsolationForest(
    n_estimators=150,
    contamination=0.03,
    random_state=42,
    n_jobs=-1
)

iso_forest.fit(train_df[num_features])

train_df['OOD_Score']=iso_forest.decision_function(train_df[num_features])
test_df['OOD_Score']=iso_forest.decision_function(test_df[num_features])

train_df['Is_OOD'] = (train_df['OOD_Score'] < 0).astype(int)
test_df['Is_OOD'] = (test_df['OOD_Score'] < 0).astype(int)

print("OOD Novelty detection report")
print(f"Train Normal In-Distribution: {(train_df['Is_OOD'] == 0).sum():,} ({(train_df['Is_OOD'] == 0).mean()*100:.1f}%)")
print(f"Train Flagged OOD Novelty:    {(train_df['Is_OOD'] == 1).sum():,} ({(train_df['Is_OOD'] == 1).mean()*100:.1f}%)")
print(f"Test Normal In-Distribution:  {(test_df['Is_OOD'] == 0).sum():,} ({(test_df['Is_OOD'] == 0).mean()*100:.1f}%)")
print(f"Test Flagged OOD Novelty:     {(test_df['Is_OOD'] == 1).sum():,} ({(test_df['Is_OOD'] == 1).mean()*100:.1f}%)")

ood_samples = test_df[test_df['Is_OOD'] == 1][num_features].describe()
print("\n=== CHARACTERISTICS OF DETECTED OOD ORDERS (TEST SET) ===")
print(ood_samples[['Sales per customer', 'Benefit per order', 'Order Item Quantity', 'Product Price']].round(2))

```

### Decision Log Entry: Outlier Treatment vs. OOD Novelty Detection
* **Decision**: Retain extreme operational shipments in the training distribution; fit an Isolation Forest strictly on training set to score operational novelty at inference time.
* **Alternative Considered**: Deleting all records with a Z-score above 3 or clipping IQR boundaries.
* **Justification**: Arbitrary deletion strips out high-value, rare shipments, distorting operational reality. An explicit OOD detector flags unfamiliar orders for human review without biasing training weights or discarding valid transactions.

```python
import numpy as np
import pandas as pd

def run_data_quality_audit(df: pd.DataFrame) -> dict:
    """
    Executes a comprehensive Trustworthy Data Quality Audit covering:
    1. Duplicate grains (Line item vs. Order level)
    2. Physical and financial value validity
    3. Cross-field semantic consistency
    4. Outlier distribution profiling
    """
    audit_results = {}
    

    exact_duplicates = df.duplicated().sum()
    item_id_duplicates = df['Order Item Id'].duplicated().sum()
    unique_orders = df['Order Id'].nunique()
    total_rows = len(df)
    items_per_order = total_rows / unique_orders
    
    audit_results['duplicates'] = {
        'exact_row_duplicates': exact_duplicates,
        'order_item_id_duplicates': item_id_duplicates,
        'unique_orders': unique_orders,
        'average_items_per_order': round(items_per_order, 2)
    }
    

    expected_total = df['Product Price'] * df['Order Item Quantity'] * (1.0 - df['Order Item Discount Rate'])
    total_discrepancies = np.abs(expected_total - df['Order Item Total']) > 0.05
    

    expected_profit_ratio = df['Benefit per order'] / df['Sales per customer']
    ratio_discrepancies = np.abs(expected_profit_ratio - df['Order Item Profit Ratio']) > 0.05
    

    negative_benefit_count = (df['Benefit per order'] < 0).sum()
    
    audit_results['consistency'] = {
        'financial_ledger_mismatches': total_discrepancies.sum(),
        'profit_ratio_discrepancies': ratio_discrepancies.sum(),
        'loss_making_orders': negative_benefit_count,
        'loss_making_order_pct': round((negative_benefit_count / total_rows) * 100, 2)
    }
    

    outlier_summary = []
    features_to_check = ['Sales per customer', 'Product Price', 'Benefit per order', 'Order Item Quantity']
    
    for col in features_to_check:
        mean_val = df[col].mean()
        std_val = df[col].std()
        z_scores = np.abs((df[col] - mean_val) / std_val)
        z3_outliers = (z_scores > 3).sum()
        
        q25, q75 = df[col].quantile(0.25), df[col].quantile(0.75)
        iqr = q75 - q25
        iqr_outliers = ((df[col] < (q25 - 1.5 * iqr)) | (df[col] > (q75 + 1.5 * iqr))).sum()
        
        outlier_summary.append({
            'Feature': col,
            'Z_Score_gt_3': z3_outliers,
            'Z_Score_pct': round((z3_outliers / total_rows) * 100, 2),
            'IQR_Outliers': iqr_outliers,
            'IQR_pct': round((iqr_outliers / total_rows) * 100, 2),
            'Min': round(df[col].min(), 2),
            'Max': round(df[col].max(), 2)
        })
        
    audit_results['outliers'] = pd.DataFrame(outlier_summary)
    return audit_results


audit = run_data_quality_audit(data_df)

print("1. Duplicate audit report")
for k, v in audit['duplicates'].items():
    print(f"{k}: {v}")

print("\n2. semantic consistency audit report")
for k, v in audit['consistency'].items():
    print(f"{k}: {v}")

print("\n3. Outliers")
print(audit['outliers'].to_string(index=False))
```

* Identifiers → candidates for removal

## Step 3 - Remove Unnecessary Features


  * **Constant features**: A feature with only 1 unique value (`Product Status`) has zero variance and provides no information.
  * **Identifier features**: Arbitrary database keys (`Customer Id`, `Order Customer Id`, `Order Item Id`) identify records but have no generalizable pattern and risk overfitting.
  * **Redundant features**: Columns that duplicate another feature (e.g., `Category Name` is 1:1 with `Category Id`; `Order Item Total` is identical to `Sales per customer`) create multicollinearity.
* **Decision**: Drop 1 constant + 3 identifiers + 7 redundant features = 11 columns in total.
* **Result**: Features reduced from **37 to 26**.

```python
# Step 3: Define and remove constant, identifier, and redundant features
constant_cols = ['Product Status']
identifier_cols = ['Customer Id', 'Order Customer Id', 'Order Item Id']
redundant_cols = [
    'Category Name', 'Product Category Id',
    'Department Name', 'Product Name',
    'Order Item Total', 'Order Profit Per Order', 'Order Item Product Price'
]

cols_to_remove = constant_cols + identifier_cols + redundant_cols

print(f'Constant features to remove ({len(constant_cols)}):   {constant_cols}')
print(f'Identifier features to remove ({len(identifier_cols)}): {identifier_cols}')
print(f'Redundant features to remove ({len(redundant_cols)}):  {redundant_cols}')

# Remove from both train and test
X_train.drop(columns=cols_to_remove, inplace=True)
X_test.drop(columns=cols_to_remove, inplace=True)

print(f'\nFeatures before removal: 37')
print(f'Features removed:        {len(cols_to_remove)}')
print(f'Features remaining:      {X_train.shape[1]} (X_train) and {X_test.shape[1]} (X_test)')
```

## Step 4 - Prepare Temporal & Rare Categories


### 4A — Temporal Feature Extraction


```python
# Step 4A: Extract temporal features from order date
for df in [X_train, X_test]:
    dt = pd.to_datetime(df['order date (DateOrders)'], format='%m/%d/%Y %H:%M')
    df['order_hour'] = dt.dt.hour
    df['order_day_of_week'] = dt.dt.dayofweek
    df['order_month'] = dt.dt.month
    df['is_weekend'] = (dt.dt.dayofweek >= 5).astype(int)
    df.drop(columns=['order date (DateOrders)'], inplace=True)

print("Extracted temporal features: ['order_hour', 'order_day_of_week', 'order_month', 'is_weekend']")
print(f'Updated feature count: X_train = {X_train.shape[1]}, X_test = {X_test.shape[1]}')
```

### 4B - Statistical Rare-Category Grouping


```python
# Step 4B: Group rare categories based on X_train frequencies only
rare_thresholds = {
    'Order Country': 0.005,  # 0.5% threshold
    'Customer City': 0.002,  # 0.2% threshold
    'Order State': 0.001,    # 0.1% threshold
    'Order City': 0.001      # 0.1% threshold
}

grouping_summary = []

for col, thresh in rare_thresholds.items():
    raw_train_count = X_train[col].nunique()
    raw_test_count = X_test[col].nunique()
    
    # 1. Count category frequencies in X_train only
    freqs = X_train[col].value_counts(normalize=True)
    frequent_cats = set(freqs[freqs >= thresh].index)
    
    # 2. Replace rare categories with 'Other_Rare' in X_train
    X_train[col] = X_train[col].apply(lambda x: x if x in frequent_cats else 'Other_Rare')
    
    # 3. Apply the same learned rule to X_test (unseen test categories map to 'Other_Rare')
    X_test[col] = X_test[col].apply(lambda x: x if x in frequent_cats else 'Other_Rare')
    
    grouping_summary.append({
        'Feature': col,
        'Threshold': f'{thresh*100:.1f}%',
        'Train Unique (Before)': raw_train_count,
        'Train Unique (After)': X_train[col].nunique(),
        'Test Unique (Before)': raw_test_count,
        'Test Unique (After)': X_test[col].nunique()
    })

display(pd.DataFrame(grouping_summary))
```

Apply two encoding strategies based on cardinality: One-Hot Encoding for low cardinality, and Target Encoding for high cardinality.

### 5A — One-Hot Encoding (Low-Cardinality Nominals)


```python
# Step 5A: One-Hot Encoding for low-cardinality features
from sklearn.preprocessing import OneHotEncoder

low_card_cols = [
    'Shipping Mode', 'Type', 'Customer Segment', 'Market',
    'Customer Country', 'Department Id', 'order_day_of_week', 'order_month'
]

# Initialize and fit encoder strictly on X_train
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
ohe.fit(X_train[low_card_cols])
ohe_feature_names = list(ohe.get_feature_names_out(low_card_cols))

# Transform both train and test
X_train_ohe = pd.DataFrame(ohe.transform(X_train[low_card_cols]), columns=ohe_feature_names, index=X_train.index)
X_test_ohe = pd.DataFrame(ohe.transform(X_test[low_card_cols]), columns=ohe_feature_names, index=X_test.index)

print(f'Features encoded:        {len(low_card_cols)}')
print(f'Binary columns created:  {len(ohe_feature_names)}')
print(f'X_train_ohe shape:       {X_train_ohe.shape}')
print(f'X_test_ohe shape:        {X_test_ohe.shape}')
```

### 5B - Regularized Target Encoding (Medium & High Cardinality)

```python
# Step 5B: 5-Fold Regularized Target Encoding for medium & high-cardinality features
from sklearn.preprocessing import TargetEncoder
from sklearn.model_selection import KFold

target_encode_cols = [
    'Category Id', 'Product Card Id', 'Order Region', 'Customer State',
    'Order Country', 'Customer City', 'Order State', 'Order City'
]

# 5-fold cross-validation to prevent training target leakage
#cv_strategy = KFold(n_splits=5, shuffle=True, random_state=42)
target_encoder = TargetEncoder(
    smooth='auto',
    cv=5,
    shuffle=True,
    random_state=42
)

# Training: fit_transform using 5-fold out-of-fold estimation
X_train_te_vals = target_encoder.fit_transform(X_train[target_encode_cols], y_train)
X_train_te = pd.DataFrame(X_train_te_vals, columns=[f'{c}_te' for c in target_encode_cols], index=X_train.index)

# Test: transform using fixed training mappings (y_test is never used)
X_test_te_vals = target_encoder.transform(X_test[target_encode_cols])
X_test_te = pd.DataFrame(X_test_te_vals, columns=[f'{c}_te' for c in target_encode_cols], index=X_test.index)

print(f'Features encoded:        {len(target_encode_cols)}')
print(f'X_train_te shape:        {X_train_te.shape}')
print(f'X_test_te shape:         {X_test_te.shape}')
print(f'Target-encoded columns:  {list(X_train_te.columns)}')
```

## Step 1 - Domain-Specific Feature Engineering

We engineer 7 new operational and contextual features across four supply chain dimensions:
1. **Financial Risk Signals:** `is_loss_making` (flags negative profit transactions) and `high_discount_flag` (orders with $\ge 15\%$ markdown).
2. **Discount Severity:** `discount_ratio` measuring discount amount relative to total gross product value.
3. **Profitability Margin:** `profit_margin_ratio` capturing order profit normalized by customer sales.
4. **Spatial Route Interaction:** `geo_coord_interaction` ($Latitude \times Longitude$) to capture regional delivery cluster coordinates.
5. **Temporal Cyclical Dynamics:** `order_hour_sin` and `order_hour_cos` representing continuous 24-hour dispatch cycles without discontinuity at midnight.
6. **Peak Traffic Indicator:** `is_peak_order_hour` marking peak transaction hours.

```python
# Step 1: Feature Engineering on train and test sets
import numpy as np
import pandas as pd

for df in [X_train_encoded, X_test_encoded]:
    # 1. Financial Risk Signals
    df['is_loss_making'] = (df['Benefit per order'] < 0).astype(int)
    df['high_discount_flag'] = (df['Order Item Discount Rate'] >= 0.15).astype(int)
    
    # 2. Relative Discount Ratio
    denom = df['Product Price'] * df['Order Item Quantity']
    df['discount_ratio'] = np.where(denom > 0, df['Order Item Discount'] / denom, 0.0)
    
    # 3. Normalized Profit Margin Ratio
    sales_denom = np.abs(df['Sales per customer']) + 1e-5
    df['profit_margin_ratio'] = np.where(df['Sales per customer'] != 0, df['Benefit per order'] / sales_denom, 0.0)
    
    # 4. Spatial Coordinate Interaction (Dispatch Location Disparity Proxy)
    df['geo_coord_interaction'] = df['Latitude'] * df['Longitude']
    
    # 5. Cyclical 24-Hour Time Features
    df['order_hour_sin'] = np.sin(2 * np.pi * df['order_hour'] / 24.0)
    df['order_hour_cos'] = np.cos(2 * np.pi * df['order_hour'] / 24.0)
    df['is_peak_order_hour'] = df['order_hour'].isin([12, 13, 14, 18, 19, 20, 21]).astype(int)

print(f"X_train_encoded shape after Feature Engineering: {X_train_encoded.shape}")
print(f"X_test_encoded shape after Feature Engineering:  {X_test_encoded.shape}")
print(f"Engineered features added (7): ['is_loss_making', 'high_discount_flag', 'discount_ratio', 'profit_margin_ratio', 'geo_coord_interaction', 'order_hour_sin', 'order_hour_cos', 'is_peak_order_hour']")
```

# Scaling, Class Imbalance & Data Splitting - IT24103276

## 1. Feature Scaling

This section investigates and prepares scaling only. Class-imbalance treatment and any new validation or data-splitting work are intentionally left for a later step.

### Existing pipeline and scaling scope

The notebook already creates `train_df` and `test_df` with `GroupShuffleSplit` (80/20, `random_state=42`) grouped by `Order Id`; the saved audit reports zero Order Id overlap. Member 2's one-hot and cross-validated target encodings are assembled into `X_train_encoded` / `X_test_encoded`, and Member 3 hands off the selected `X_train_final` / `X_test_final` matrices (74 features) with `y_train` / `y_test`. The selected matrices are numeric, but include both actual numeric measures and numeric representations of categorical/indicator features. No scaler is already applied.

Scaling is model-dependent. Logistic Regression, KNN, SVM, and neural-network models are generally sensitive to feature magnitude, so scaling their numeric inputs is useful. Decision Trees, Random Forests, and gradient-boosted trees generally do not require scaled features. The notebook has not selected a final classifier; the Random Forest in the feature-selection section is for feature importance, not a final evaluated classifier. Therefore this section creates a separate scaled feature path for scale-sensitive models and leaves the existing unscaled matrices untouched for tree-based models.

The explicit whitelist below uses only operational numeric fields that remain in Member 3's selected feature matrix. One-hot columns, target-encoded columns, binary indicators, cyclical hour features, the target, identifiers, and excluded/leakage columns are not scaled.

```python
import numpy as np
import pandas as pd

# These are the unencoded numeric measures retained in the current final feature handoff.
scale_feature_candidates = [
    'Days for shipment (scheduled)',
    'Benefit per order',
    'Sales per customer',
    'Latitude',
    'Longitude',
    'Order Item Discount',
    'Order Item Discount Rate',
    'Order Item Profit Ratio',
    'Order Item Quantity',
    'Product Price',
    'order_hour',
    'geo_coord_interaction',
]

assert X_train_final.columns.equals(X_test_final.columns), 'Train/test feature columns must match.'
missing_scale_features = [c for c in scale_feature_candidates if c not in X_train_final.columns]
if missing_scale_features:
    raise ValueError(f'Expected selected numeric features are missing: {missing_scale_features}')

scale_feature_cols = scale_feature_candidates
assert target_col not in scale_feature_cols
assert 'Order Id' not in scale_feature_cols
assert not set(scale_feature_cols).intersection(leakage_cols)

train_numeric = X_train_final[scale_feature_cols]
q1 = train_numeric.quantile(0.25)
q3 = train_numeric.quantile(0.75)
iqr = q3 - q1
iqr_outlier_pct = (
    train_numeric.lt(q1 - 1.5 * iqr) | train_numeric.gt(q3 + 1.5 * iqr)
).mean() * 100

scale_profile = train_numeric.agg(['min', 'max', 'mean', 'std']).T
scale_profile['IQR outliers (%)'] = iqr_outlier_pct
print(f'Features to scale ({len(scale_feature_cols)}): {scale_feature_cols}')
display(scale_profile.round(3))
```

### Scaler choice

The training-partition profile shows materially different ranges (for example, `Product Price` and `geo_coord_interaction` span thousands, while `Order Item Discount Rate` is bounded between 0 and 0.25). It also shows IQR-defined extreme values in operational fields, including `Benefit per order`, `Order Item Profit Ratio`, and `Order Item Discount`. This supports `RobustScaler`: it centers by the training median and scales by the training interquartile range, making the scale less sensitive to extreme observations than `StandardScaler`. It does not remove, clip, or otherwise alter outliers. The code computes the summary from the selected training features so the evidence stays tied to the current pipeline.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler

# Keep this as a separate model-specific path; do not overwrite the unscaled handoff.
scale_sensitive_pipeline = Pipeline(steps=[
    ('column_transformer', ColumnTransformer(
        transformers=[('robust_scale', RobustScaler(), scale_feature_cols)],
        remainder='passthrough',
        verbose_feature_names_out=False,
    ))
])

# Fit only on the existing training partition; apply the same fitted transform to test.
X_train_scaled = pd.DataFrame(
    scale_sensitive_pipeline.fit_transform(X_train_final),
    columns=scale_sensitive_pipeline.get_feature_names_out(),
    index=X_train_final.index,
)
X_test_scaled = pd.DataFrame(
    scale_sensitive_pipeline.transform(X_test_final),
    columns=scale_sensitive_pipeline.get_feature_names_out(),
    index=X_test_final.index,
)

fitted_scaler = scale_sensitive_pipeline.named_steps['column_transformer'].named_transformers_['robust_scale']
training_iqr = (
    X_train_final[scale_feature_cols].quantile(0.75)
    - X_train_final[scale_feature_cols].quantile(0.25)
).to_numpy()
expected_robust_scale = np.where(training_iqr == 0, 1.0, training_iqr)
assert np.allclose(fitted_scaler.center_, X_train_final[scale_feature_cols].median().to_numpy())
assert np.allclose(fitted_scaler.scale_, expected_robust_scale)

passthrough_cols = [c for c in X_train_final.columns if c not in scale_feature_cols]
assert set(X_train_scaled.columns) == set(X_train_final.columns)
assert set(X_test_scaled.columns) == set(X_test_final.columns)
assert np.array_equal(X_train_scaled[passthrough_cols].to_numpy(), X_train_final[passthrough_cols].to_numpy())
assert np.array_equal(X_test_scaled[passthrough_cols].to_numpy(), X_test_final[passthrough_cols].to_numpy())
assert not X_train_scaled.isna().any().any() and not X_test_scaled.isna().any().any()
assert target_col not in X_train_scaled.columns and 'Order Id' not in X_train_scaled.columns

print('Scaler:', type(fitted_scaler).__name__)
print('Fit rows:', len(X_train_final), '| transformed test rows:', len(X_test_final))
print('Fit/transform leakage check: PASS (fit on X_train_final only; X_test_final transformed only)')
print('Training median / IQR match check: PASS')
print(f'Unscaled passthrough columns: {len(passthrough_cols)} (encoded categorical and bounded indicator/cyclical features)')
print('Output shapes:', X_train_scaled.shape, X_test_scaled.shape)
display(X_train_scaled[scale_feature_cols].agg(['min', 'max', 'mean', 'std', 'median']).T.round(3))
```

### Scaling Decision

Scaling is prepared for scale-sensitive model families because the selected operational numeric fields have very different units and ranges. The section scales only the 12 unencoded numeric measures listed in the profile, using `RobustScaler` because the training data contains IQR-defined extreme observations; those observations are retained. Member 2's one-hot and target-encoded features, binary indicators, cyclical hour encodings, and the target are passed through or excluded from the scaling columns. Logistic Regression, KNN, SVM, and neural-network models can use `X_train_scaled` / `X_test_scaled`; tree-based models can continue using the original `X_train_final` / `X_test_final`.

No new split is created. The existing `GroupShuffleSplit` by `Order Id` is reused: the pipeline is fitted with `X_train_final` only, then the same fitted transformer is applied to `X_test_final`. Assertions verify the scaler's medians and IQRs match training-only statistics and that passthrough features remain unchanged. Class-imbalance handling and validation strategy are not implemented in this section.

### 3. Data Splitting & Validation

This subsection validates the existing grouped split instead of creating a second split. The notebook already uses `GroupShuffleSplit` with `train_size=0.8`, `random_state=42`, and groups by `Order Id`, which is the correct leakage-safe approach because multiple rows may belong to the same order.

The goal is not to replace a correct split, but to audit it, confirm the proportion, and create a separate grouped cross-validation strategy using the training partition only for later model selection.

#### Why grouping by `Order Id` is necessary

A random row-level split is inappropriate here because several records can share the same `Order Id` (for example, different line items from one purchase order). Grouping by `Order Id` prevents this by guaranteeing that each order is assigned to only one side of the split.

### Decision Log — SGD-SVM

**Model choice**: `SGDClassifier` was selected over kernel `SVC` because `X_train_scaled` contains 144,650 rows. Kernel SVM has $O(n^2)$ memory and $O(n^3)$ training complexity, making it infeasible at this scale. `SGDClassifier(loss='hinge')` implements the same maximum-margin objective as a linear SVM but trains via stochastic gradient descent with $O(n)$ complexity per epoch.

**Loss functions in grid**: `hinge` (true SVM margin loss), `modified_huber` (smooth approximation, supports `predict_proba`), `log_loss` (logistic regression within the same SGD framework — included for direct comparison).

**Probability calibration**: If the best loss is `hinge`, a `CalibratedClassifierCV` (Platt scaling, 3-fold) wraps the base estimator to enable `predict_proba` for log loss and ROC-AUC on the test set.

**Scaling**: `X_train_scaled` / `X_test_scaled` from the upstream `RobustScaler` pipeline were used directly. SGD convergence speed and weight regularisation are sensitive to feature scale.

**Leakage prevention**: `GroupKFold(n_splits=5)` grouped by `Order Id` was passed to `GridSearchCV` as a pre-generated split list, ensuring no order's rows appear in both the training and validation fold of any CV iteration.

**Class imbalance**: `class_weight='balanced'` was applied in both the grid search and the final fitted model to up-weight the minority class (`0 = On-Time`, 45.17%).
